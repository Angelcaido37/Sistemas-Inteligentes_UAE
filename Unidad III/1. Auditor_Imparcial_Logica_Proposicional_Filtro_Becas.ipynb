# ⚖️ El Auditor Imparcial: Lógica Proposicional para Decisiones sin Sesgo

## 🎯 Objetivo
Traducir lenguaje natural (ambiguo) a **proposiciones lógicas matemáticas** programables y evaluar reglas de decisión en milisegundos.

**Reto social:** crear un sistema que reduzca el sesgo humano aplicando reglas explícitas y verificables (aunque eso también trae dilemas de justicia y rigidez).


## 🧠 1) INICIO — Abstracción (Matemáticas)

### Dinámica: *El Abogado del Diablo*
Se presentan sentencias legales ambiguas, por ejemplo:

> “Se prohíbe el paso a vehículos grandes.”

❓ Preguntas:
- ¿Qué es “grande”? (¿peso, altura, número de ejes, volumen?)
- ¿Se aplica a ambulancias?
- ¿Cómo se verifica?

### Concepto: Formalización
Para programar una regla, necesitamos convertirla a lógica:

- **Proposición:** enunciado que puede ser Verdadero (V) o Falso (F)
- **Implicación:** \(P \rightarrow Q\)
- **Conectores:** \(\land\) (y), \(\lor\) (o), \(\neg\) (no)

Las **tablas de verdad** permiten validar reglas antes de programarlas.


## 🧩 2) DESARROLLO — Programación (Tecnología)

### Actividad: Programar un *Filtro de Becas*

Definimos proposiciones para cada candidato:

- **P:** Vive en zona rural  
- **Q:** Promedio \(> 9.0\)  
- **R:** Ingreso \(< 5000\)  

**Regla de aprobación:**

\[
(((P∨R)∧Q)→A)
\]

📌 Interpretación:
- Para aprobar, el candidato debe tener **promedio alto (Q)** y además cumplir **al menos una** condición de necesidad/contexto: **rural (P) o ingreso bajo (R)**.


## ✅ Tabla de verdad (validación de la regla)
Vamos a generar automáticamente la tabla de verdad para \((P\lor R)\land Q\) y decidir Aprobado.


```python
import pandas as pd

def regla_aprobado(P, Q, R):
    # (P ∨ R) ∧ Q
    return (P or R) and Q

rows = []
for P in [False, True]:
    for Q in [False, True]:
        for R in [False, True]:
            A = regla_aprobado(P, Q, R)
            rows.append({
                "P (Rural)": P,
                "Q (Prom>9)": Q,
                "R (Ing<5000)": R,
                "Aprobado": A
            })

tabla = pd.DataFrame(rows)
tabla
```

### 🔎 ¿Qué estamos comprobando?
- Si **Q es falso** (promedio no supera 9.0), **nunca aprueba** (aunque haya necesidad).
- Si **Q es verdadero** y **P o R** es verdadero, entonces aprueba.

Esto hace que la regla sea **clara**, pero puede ser **rígida** (lo discutimos al final).


## 👩‍💻 Código en vivo: Evaluar 100 candidatos simulados

Simularemos una lista de candidatos con:
- `rural` (Sí/No)
- `promedio` (0 a 10)
- `ingreso` (0 a 12000)

Luego traducimos eso a P, Q, R y aplicamos la regla.


```python
import random
import pandas as pd

random.seed(42)

def simular_candidatos(n=100):
    data = []
    for i in range(1, n + 1):
        rural = random.random() < 0.35  # 35% zona rural (simulado)
        promedio = round(random.uniform(7.0, 10.0), 2)
        ingreso = random.randint(1500, 12000)
        data.append({
            "id": i,
            "rural": rural,
            "promedio": promedio,
            "ingreso": ingreso
        })
    return pd.DataFrame(data)

candidatos = simular_candidatos(100)

def evaluar(df):
    df = df.copy()
    df["P_rural"] = df["rural"]
    df["Q_prom_mayor_9"] = df["promedio"] > 9.0
    df["R_ingreso_menor_5000"] = df["ingreso"] < 5000
    df["aprobado"] = (df["P_rural"] | df["R_ingreso_menor_5000"]) & df["Q_prom_mayor_9"]
    return df

resultado = evaluar(candidatos)
resultado.head(10)
```

### 📊 Resumen rápido


```python
resultado["aprobado"].value_counts()
```

### Casos frontera (para discusión de justicia/rigidez)


```python
frontera = resultado[(resultado["promedio"] >= 8.95) & (resultado["promedio"] <= 9.05)].sort_values("promedio")
frontera.head(15)
```

## 🧠 3) CIERRE — Reflexión crítica

### ¿Es justo el algoritmo?
- ✔️ Ventaja: reglas explícitas, auditables, sin “favoritismos” personales.
- ❌ Riesgo: rigidez. Ejemplo: promedio **8.99** con necesidad extrema → **no aprueba**.

### Discusión guiada
- ¿Deberíamos permitir una “zona gris” (ej. 8.9–9.0) con revisión humana?
- ¿Qué variables faltan? (discapacidad, distancia a escuela, responsabilidades familiares, etc.)
- ¿Cómo balancear **equidad** vs **transparencia**?

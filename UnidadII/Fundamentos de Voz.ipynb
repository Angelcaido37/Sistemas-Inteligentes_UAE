## 1. Reconocimiento de Voz (SpeechRecognition)

Usaremos la librería `speech_recognition` que actúa como un wrapper para varias APIs (Google, Bing, Sphinx).

**Nota:** El reconocimiento de Google requiere internet.

```python
import speech_recognition as sr

def reconocer_voz():
    # 1. Crear el objeto reconocedor
    recognizer = sr.Recognizer()
    
    # Ajustar umbral de energía (opcional, para ambientes ruidosos)
    recognizer.energy_threshold = 4000

    # 2. Usar el micrófono como fuente
    try:
        with sr.Microphone() as source:
            print("🎤 Ajustando ruido ambiente... espera un segundo.")
            recognizer.adjust_for_ambient_noise(source, duration=1)
            
            print("🗣️ ¡Habla ahora!")
            audio = recognizer.listen(source, timeout=5)
            
            print("⏳ Procesando...")
            
            # 3. Enviar a la API de Google
            texto = recognizer.recognize_google(audio, language='es-ES')
            return texto
            
    except sr.WaitTimeoutError:
        return "No escuché nada."
    except sr.UnknownValueError:
        return "No entendí lo que dijiste."
    except sr.RequestError as e:
        return f"Error de conexión: {e}"
    except Exception as e:
        return f"Error de micrófono: {e}"

# Probamos la función
resultado = reconocer_voz()
print(f"📝 Resultado: {resultado}")
```

## 2. Síntesis de Voz (Text-to-Speech)

Existen dos enfoques principales:
1.  **Online (gTTS):** Mejor calidad, voces de Google, genera MP3.
2.  **Offline (pyttsx3):** Más rápido, voces robóticas del sistema, no requiere internet.

### 2.1 Enfoque Online: gTTS

```python
from gtts import gTTS
import os
import time

def hablar_online(texto):
    print(f"🔊 Generando audio online: {texto}")
    tts = gTTS(text=texto, lang='es', slow=False)
    archivo = "salida_gtts.mp3"
    tts.save(archivo)
    
    # Reproducir según el sistema operativo
    # Windows: start, Mac: afplay, Linux: mpg321 o xdg-open
    if os.name == 'nt': # Windows
        os.system(f'start {archivo}')
    else: 
        os.system(f'afplay {archivo}') # Mac/Linux (ajustar comando)

hablar_online("Hola, esto es una prueba usando los servidores de Google.")
```

### 2.2 Enfoque Offline: pyttsx3
Este es el que usaremos preferentemente para asistentes virtuales por su velocidad de respuesta.

```python
import pyttsx3

def inicializar_motor():
    engine = pyttsx3.init()
    
    # Configuración
    engine.setProperty('rate', 150)    # Velocidad
    engine.setProperty('volume', 0.9)  # Volumen
    
    # Listar voces disponibles e intentar poner español
    voices = engine.getProperty('voices')
    for voice in voices:
        # Buscamos alguna voz en español en el sistema
        if "spanish" in voice.name.lower() or "es" in voice.languages:
            engine.setProperty('voice', voice.id)
            break
    
    return engine

def hablar_offline(texto):
    engine = inicializar_motor()
    print(f"🤖 Sistema: {texto}")
    engine.say(texto)
    engine.runAndWait()

hablar_offline("Hola, soy tu sistema offline. Respondo mucho más rápido.")
```

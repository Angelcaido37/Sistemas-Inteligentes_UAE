## 🔊 Configuración de la voz (Actuador)
En este ejemplo usamos `pyttsx3`, que funciona sin conexión.

- `rate`: velocidad de habla (más bajo = más lento)
- `voice`: tipo de voz (depende del sistema operativo)


````python
import pyttsx3
import speech_recognition as sr

# Configuración de la voz
engine = pyttsx3.init()
engine.setProperty('rate', 140)  # Velocidad moderada

def hablar(texto):
    """Actuador: convierte texto en voz"""
    engine.say(texto)
    engine.runAndWait()
````

## 👂 Función de escucha (Sensor) — versión resumida
📌 En clase puedes reutilizar la función `escuchar_usuario()` del notebook anterior.

Aquí dejamos una versión compacta y robusta, con calibración de ruido y manejo de errores.


````python
def escuchar_usuario(timeout=5, language="es-MX"):
    """Sensor: escucha al usuario y devuelve texto en minúsculas."""
    r = sr.Recognizer()
    with sr.Microphone() as source:
        print("🎧 Calibrando ruido de fondo... (silencio)")
        r.adjust_for_ambient_noise(source, duration=1)
        print("🗣️ Escuchando...")
        try:
            audio = r.listen(source, timeout=timeout)
            texto = r.recognize_google(audio, language=language)
            print(f"Usuario dijo: {texto}")
            return texto.lower()
        except sr.UnknownValueError:
            print("❌ No entendí el audio.")
            return ""
        except sr.RequestError:
            print("⚠️ Error de conexión con el servicio.")
            return ""
        except sr.WaitTimeoutError:
            print("⌛ Tiempo de espera agotado.")
            return 
````

## 🏥 Kiosco Inteligente (Agente completo)
Este agente:
1. **Habla** para iniciar la interacción (TTS)
2. **Escucha** al usuario (STT)
3. **Decide** por palabras clave
4. **Responde** con voz (TTS)

📌 En este notebook incluimos una **simulación** y una **versión real** con micrófono.


### ✅ Versión simulada (para probar sin micrófono)


````python
def kiosco_inteligente_simulado():
    hablar("Bienvenido al sistema de salud. ¿En qué puedo ayudarle?")

    # Simulación (esto vendría del micrófono)
    entrada_usuario = "Quiero una cita médica"
    print(f"(Simulado) Usuario: {entrada_usuario}")
    entrada_usuario = entrada_usuario.lower()

    if "cita" in entrada_usuario:
        hablar("Entendido. Abriendo agenda de doctores.")
    elif "farmacia" in entrada_usuario:
        hablar("La farmacia está en el pasillo dos.")
    else:
        hablar("Lo siento, no comprendí su solicitud.")

# Ejecutar simulación
kiosco_inteligente_simulado()
````

### 🎙️ Versión real (usa micrófono)


````python
def kiosco_inteligente_real():
    hablar("Bienvenido al sistema de salud. ¿En qué puedo ayudarle?")

    entrada_usuario = escuchar_usuario()

    if "cita" in entrada_usuario:
        hablar("Entendido. Abriendo agenda de doctores.")
    elif "farmacia" in entrada_usuario:
        hablar("La farmacia está en el pasillo dos.")
    elif "ayuda" in entrada_usuario or "socorro" in entrada_usuario:
        hablar("Entendido. Voy a solicitar ayuda inmediatamente.")
    else:
        hablar("Lo siento, no comprendí su solicitud.")

# ⚠️ Descomenta para ejecutar con micrófono:
# kiosco_inteligente_real()
````

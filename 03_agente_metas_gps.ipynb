# 03 — Agente basado en metas (GPS con búsqueda)

**Idea clave:** un agente *basado en metas* decide acciones para **alcanzar un objetivo**.
Ejemplo típico: un **GPS** que calcula una ruta del punto A al B.

Aquí representamos lugares como **nodos** y calles como **aristas con costo**.
Luego usamos **Dijkstra** para encontrar el camino de costo mínimo.


```python
# ============================================================
# AGENTE BASADO EN METAS: GPS (BÚSQUEDA DE RUTA)
# ============================================================

import numpy as np

# 1) Grafo: nodo -> vecinos (con costo)
grafo = {
    "Casa": {"Tienda": 4, "Parque": 2},
    "Parque": {"Casa": 2, "Escuela": 5, "Tienda": 3},
    "Tienda": {"Casa": 4, "Parque": 3, "Trabajo": 6},
    "Escuela": {"Parque": 5, "Trabajo": 4},
    "Trabajo": {"Tienda": 6, "Escuela": 4}
}

inicio = "Casa"
meta = "Trabajo"

# 2) Dijkstra (sin librerías externas)
def dijkstra(grafo, inicio, meta):
    dist = {nodo: float("inf") for nodo in grafo}  # distancias
    dist[inicio] = 0

    prev = {nodo: None for nodo in grafo}          # para reconstruir ruta
    no_visitados = set(grafo.keys())               # nodos por visitar

    while no_visitados:
        # a) nodo con menor distancia conocida
        actual = min(no_visitados, key=lambda n: dist[n])
        no_visitados.remove(actual)

        # b) si llegamos a la meta, paramos
        if actual == meta:
            break

        # c) “relajación” de vecinos
        for vecino, costo in grafo[actual].items():
            if vecino in no_visitados:
                nueva_dist = dist[actual] + costo
                if nueva_dist < dist[vecino]:
                    dist[vecino] = nueva_dist
                    prev[vecino] = actual

    # d) reconstrucción del camino
    ruta = []
    nodo = meta
    while nodo is not None:
        ruta.append(nodo)
        nodo = prev[nodo]
    ruta.reverse()

    return ruta, dist[meta]

ruta, costo_total = dijkstra(grafo, inicio, meta)

print("Ruta encontrada:", " -> ".join(ruta))
print("Costo total:", costo_total)
```

## ¿Cómo se aplica en industria?
- Rutas de reparto/logística.
- Secuencias en almacenes (picking).
- Planeación de mantenimiento (visitar puntos con menor costo/tiempo).

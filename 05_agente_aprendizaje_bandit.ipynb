# 05 — Agente que aprende (multi-armed bandit)

**Idea clave:** el agente **mejora con experiencia**.
Ejemplo típico: probar diferentes campañas/promociones y aprender cuál funciona mejor.

Usamos una estrategia simple llamada **epsilon-greedy**:
- con probabilidad ε exploro (pruebo al azar),
- con probabilidad 1-ε exploto (elijo la mejor estimada).


```python
# ============================================================
# AGENTE QUE APRENDE: MULTI-ARMED BANDIT (E-GREEDY)
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

# 1) Probabilidades reales (desconocidas para el agente)
p_real = {"A": 0.05, "B": 0.08, "C": 0.03}  # tasa real de éxito

opciones = list(p_real.keys())

# 2) Parámetros
N = 2000
epsilon = 0.10

# 3) Memoria del agente
conteos = {k: 0 for k in opciones}
exitos  = {k: 0 for k in opciones}
estimado = {k: 0.0 for k in opciones}

hist_mejor = []
hist_reward = []

for _ in range(N):
    # ---- ELEGIR ----
    if np.random.rand() < epsilon:
        eleccion = np.random.choice(opciones)          # explorar
    else:
        eleccion = max(estimado, key=estimado.get)     # explotar

    # ---- ACTUAR ----
    reward = 1 if np.random.rand() < p_real[eleccion] else 0

    # ---- CRÍTICO: registrar ----
    conteos[eleccion] += 1
    exitos[eleccion] += reward

    # ---- APRENDER: actualizar estimación ----
    estimado[eleccion] = exitos[eleccion] / conteos[eleccion]

    hist_mejor.append(max(estimado, key=estimado.get))
    hist_reward.append(reward)

print("Estimaciones finales:")
for k in opciones:
    print(f"  {k}: estimado={estimado[k]:.3f} | real={p_real[k]:.3f} | intentos={conteos[k]}")

print("\nOpción elegida como mejor:", max(estimado, key=estimado.get))

# Éxito promedio acumulado
reward_prom = np.cumsum(hist_reward) / (np.arange(N) + 1)

plt.figure(figsize=(10,4))
plt.plot(reward_prom)
plt.title("Aprendizaje: éxito promedio acumulado")
plt.xlabel("Iteración")
plt.ylabel("Éxito promedio")
plt.show()
```

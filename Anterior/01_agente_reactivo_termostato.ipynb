# 01 — Agente reactivo simple (Termostato)

**Idea clave (IA basada en reglas):** un *agente reactivo simple* aplica reglas del tipo **SI condición ENTONCES acción**.
No usa memoria, decide con lo que *percibe en este instante*.

## Caso vida diaria
Un **termostato**:
- **Sensor (percepción):** temperatura actual.
- **Actuador (acción):** encender/apagar calefacción.
- **Reglas:** si hace frío -> encender; si hace calor -> apagar.

Abajo simulamos 24 horas, como si fueran lecturas de un sensor real.


```python
# ============================================================
# AGENTE REACTIVO SIMPLE: TERMOSTATO
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

# 1) Definimos reglas del termostato (zona de confort)
TEMP_MIN = 20  # Si baja de 20°C, encendemos
TEMP_MAX = 22  # Si sube de 22°C, apagamos

# 2) Simulamos temperaturas de un día (24 horas)
#    Usamos una señal base + ruido para que parezca real.
horas = np.arange(24)                                       # horas 0..23
temperatura = 21 + 2*np.sin(2*np.pi*horas/24)              # patrón diario
temperatura = temperatura + np.random.normal(0, 0.4, 24)   # ruido realista

# 3) El "estado" del actuador: calefacción encendida o apagada
calefaccion_encendida = False

# 4) Guardamos decisiones para analizarlas después
acciones = []  # "ENCENDER" / "APAGAR" / "MANTENER"
estados = []   # True/False para ver si quedó encendida o no

# 5) Ciclo percepción -> decisión -> acción (hora por hora)
for t in temperatura:
    # ---- PERCEPCIÓN ----
    temp_actual = t

    # ---- REGLAS (SI... ENTONCES...) ----
    if temp_actual < TEMP_MIN:
        calefaccion_encendida = True
        acciones.append("ENCENDER")
    elif temp_actual > TEMP_MAX:
        calefaccion_encendida = False
        acciones.append("APAGAR")
    else:
        acciones.append("MANTENER")

    # ---- GUARDAMOS ESTADO ----
    estados.append(calefaccion_encendida)

# 6) Mostramos resultados en consola (resumen)
for h in range(24):
    print(f"Hora {h:02d}: Temp={temperatura[h]:.1f}°C -> Acción={acciones[h]} -> Calefacción={estados[h]}")

# 7) Graficamos temperatura y estado de calefacción
plt.figure(figsize=(10,4))
plt.plot(horas, temperatura, marker="o")
plt.axhline(TEMP_MIN, linestyle="--")
plt.axhline(TEMP_MAX, linestyle="--")
plt.title("Termostato reactivo simple (sin memoria)")
plt.xlabel("Hora")
plt.ylabel("Temperatura (°C)")
plt.show()

plt.figure(figsize=(10,2))
plt.step(horas, estados, where="mid")
plt.title("Estado del actuador (calefacción)")
plt.xlabel("Hora")
plt.ylabel("Encendida (1) / Apagada (0)")
plt.ylim(-0.1, 1.1)
plt.show()
```

## ¿Cómo se aplica en industria?
Este patrón es típico en control **ON/OFF**:
- refrigeración de cámaras frías,
- bombas que se activan si baja el nivel,
- ventilación automática por temperatura,
- alarmas por umbral.

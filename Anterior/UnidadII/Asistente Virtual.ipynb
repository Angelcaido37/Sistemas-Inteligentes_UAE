## Agentes Inteligentes y Accesibilidad

Vamos a construir un asistente completo (tipo Siri o Alexa básico). Este agente será capaz de realizar acciones reales en el sistema operativo.

### ♿ Accesibilidad y Usabilidad
Las interfaces de voz son cruciales para:
* **Discapacidad Visual:** Permite usar la PC sin pantalla.
* **Discapacidad Motriz:** Control "manos libres".
 
**Principios de diseño que aplicaremos:**
1.  **Feedback Auditivo:** Confirmar siempre que se entendió el comando.
2.  **Manejo de Errores:** No fallar silenciosamente, informar al usuario.

```python
import speech_recognition as sr
import pyttsx3
import datetime
import webbrowser # Para abrir URLs
import os
import time
```

## 1. Clase Asistente Virtual

Utilizaremos un diseño modular. Cada comando será una función separada mapeada en un diccionario. Esto facilita escalar el sistema.

```python
class AsistenteVirtual:
    def __init__(self):
        # Inicialización de componentes
        self.recognizer = sr.Recognizer()
        self.engine = pyttsx3.init()
        self.engine.setProperty('rate', 150)
        
        # Mapeo de comandos a funciones (Sistemas Simbólicos)
        # La clave es la palabra que activa la función
        self.comandos = {
            'hora': self.decir_hora,
            'fecha': self.decir_fecha,
            'buscar': self.buscar_web,
            'ayuda': self.mostrar_ayuda,
            'salir': self.salir
        }
        
        self.activo = True

    def hablar(self, texto):
        print(f"🤖 Asistente: {texto}")
        self.engine.say(texto)
        self.engine.runAndWait()

    def escuchar(self):
        with sr.Microphone() as source:
            print("🎤 Escuchando...")
            self.recognizer.adjust_for_ambient_noise(source, duration=0.5)
            try:
                # Timeout corto para que sea ágil
                audio = self.recognizer.listen(source, timeout=5)
                texto = self.recognizer.recognize_google(audio, language='es-ES')
                print(f"👤 Tú: {texto}")
                return texto.lower()
            except sr.WaitTimeoutError:
                return ""
            except sr.UnknownValueError:
                self.hablar("No te entendí, ¿puedes repetir?")
                return ""
            except Exception as e:
                print(f"Error: {e}")
                return ""

    # --- FUNCIONES DE COMANDO ---
    
    def decir_hora(self, args):
        hora = datetime.datetime.now().strftime('%H:%M')
        self.hablar(f"Son las {hora}")

    def decir_fecha(self, args):
        # Formateo simple de fecha
        fecha = datetime.datetime.now().strftime('%d-%m-%Y')
        self.hablar(f"La fecha de hoy es {fecha}")

    def buscar_web(self, termino):
        if termino:
            self.hablar(f"Buscando {termino} en Google")
            webbrowser.open(f'https://www.google.com/search?q={termino}')
        else:
            self.hablar("¿Qué quieres que busque?")

    def mostrar_ayuda(self, args):
        comandos_texto = ", ".join(self.comandos.keys())
        self.hablar(f"Puedes pedirme: {comandos_texto}")

    def salir(self, args):
        self.hablar("Hasta luego. Apagando sistemas.")
        self.activo = False

    # --- MOTOR DE PROCESAMIENTO ---

    def procesar_comando(self, texto):
        # Lógica de coincidencia de comandos
        comando_encontrado = False
        
        for palabra_clave, funcion in self.comandos.items():
            if palabra_clave in texto:
                # Extraemos el argumento (lo que sigue al comando)
                # Ej: "buscar gatos" -> argumento: "gatos"
                argumento = texto.replace(palabra_clave, "").strip()
                funcion(argumento)
                comando_encontrado = True
                break
        
        if not comando_encontrado and texto != "":
            self.hablar("No reconozco ese comando. Di 'ayuda' para ver opciones.")

    def ejecutar(self):
        self.hablar("Hola, sistema iniciado. Te escucho.")
        while self.activo:
            texto = self.escuchar()
            if texto:
                self.procesar_comando(texto)
```

## 2. Iniciar Asistente
Prueba comandos como:

* "**Salir**"

```python
jarvis = AsistenteVirtual()
jarvis.ejecutar()
```

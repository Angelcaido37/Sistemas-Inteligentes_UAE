## 👂 Implementando el Oído Digital (Sensor de Audio)
Este sensor:
- Escucha desde el micrófono
- Se calibra al ruido ambiental
- Convierte voz en texto


````python
import speech_recognition as sr

def escuchar_usuario():
    r = sr.Recognizer()
    with sr.Microphone() as source:
        print("🎧 Calibrando ruido de fondo... (silencio)")
        r.adjust_for_ambient_noise(source, duration=1)
        print("🗣️ Escuchando... (Di 'Ayuda' o 'Información')")

        try:
            # Reconocimiento usando API de Google
            audio = r.listen(source, timeout=5)
            texto = r.recognize_google(audio, language="es-MX")
            print(f"Usuario dijo: {texto}")
            return texto.lower()

        except sr.UnknownValueError:
            print("❌ No entendí el audio.")
            return ""

        except sr.RequestError:
            print("⚠️ Error de conexión con el servicio.")
            return 
````

## 🧠 Lógica de decisión basada en lenguaje natural
El agente **no entiende frases completas**, solo busca **palabras clave**.


````python
# Ejecución del sensor
comando = escuchar_usuario()

# Decisión del agente
if "ayuda" in comando or "socorro" in comando:
    print("🚨 ACCIÓN: LLAMANDO AL 911...")
elif "información" in comando:
    print("ℹ️ ACCIÓN: Abriendo menú de trámites.")
else:
    print("🤖 ACCIÓN: Sin comando reconocido.")
````

# 🤖 Agente Racional con Estado (Memoria)

## 🎯 Concepto clave
Un **agente racional** hace *"lo correcto"* con la información que tiene disponible,  
no necesariamente *"lo perfecto"*.

En Inteligencia Artificial, la racionalidad depende de:
- Los **sensores** disponibles
- El **estado interno (memoria)**
- Las **reglas de decisión**


## ⚖️ Debate ético
**Situación:**  
Si el *ReciclaBot* tritura por error un **anillo de diamantes** que cayó en la basura:

> ❓ ¿Actuó racionalmente?

✔️ **Sí**, porque:
- Sus sensores **no detectan diamantes**
- Actuó correctamente con la información disponible

📌 *La IA no es moral ni inmoral: es limitada por su diseño.*


## 🧠 Agentes con Estado
A diferencia de un agente reactivo simple, un **agente con estado**:

- Mantiene **memoria interna**
- Toma decisiones considerando el **pasado**
- Puede modificar su comportamiento según su estado


## 📦 Diseño del agente
El agente ahora debe:
1. Recordar cuánta basura ha procesado
2. Saber si el contenedor está lleno
3. Solicitar vaciado cuando se alcanza la capacidad


```python
class AgenteConMemoria:
    def __init__(self):
        # Capacidad máxima del contenedor
        self.capacidad_maxima = 3

        # Estado interno (Memoria)
        self.contador_basura = 0

    def actuar(self, percepcion):
        # 1. Revisión del estado interno
        if self.contador_basura >= self.capacidad_maxima:
            print("⚠️ CONTENEDOR LLENO. Acción: SOLICITAR VACIADO.")
            self.contador_basura = 0  # Reset después de vaciar
            return

        # 2. Procesamiento de la percepción
        material = percepcion
        if material == "pet":
            print(f"Procesando {material}... Acción: RECICLAR")
            self.contador_basura += 1
        else:
            print(f"Procesando {material}... Acción: DESCARTAR")
```

## 🔁 Simulación del entorno
Simulamos una secuencia de residuos que llegan al sistema.
Observa cómo el agente **recuerda** cuántos objetos ha procesado.


```python
# Simulación
robot = AgenteConMemoria()
basuras = ["pet", "pet", "papel", "pet", "pet"]

for b in basuras:
    robot.actuar(b)
```

## 🧩 Análisis del comportamiento
- El agente **no se desborda**
- Detiene el proceso cuando el contenedor está lleno
- Usa su **memoria interna** para decidir

📌 Este es un **agente racional con estado**, no reactivo simple.

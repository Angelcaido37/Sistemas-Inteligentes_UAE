# 04 — Agente basado en utilidad (decisiones con trade-offs)

**Idea clave:** cuando hay varias opciones “buenas”, usamos una **función de utilidad**
para elegir lo **mejor** según prioridades.

## Caso empresa
Elegir ruta de entrega considerando:
- **tiempo**,
- **costo**,
- **riesgo**.

El negocio asigna pesos (qué importa más) y el agente elige.


```python
# ============================================================
# AGENTE BASADO EN UTILIDAD: ELEGIR MEJOR OPCIÓN
# ============================================================

import numpy as np

# 1) Rutas con métricas (ejemplo)
rutas = {
    "A": {"tiempo_min": 35, "costo": 60, "riesgo": 0.35},  # rápida pero riesgosa
    "B": {"tiempo_min": 45, "costo": 70, "riesgo": 0.10},  # segura
    "C": {"tiempo_min": 55, "costo": 40, "riesgo": 0.20},  # barata
}

# 2) Pesos (preferencias)
w_tiempo = 0.5
w_costo  = 0.3
w_riesgo = 0.2

# 3) Normalización a 0..1
tiempos = np.array([r["tiempo_min"] for r in rutas.values()])
costos  = np.array([r["costo"] for r in rutas.values()])
riesgos = np.array([r["riesgo"] for r in rutas.values()])

def normalizar(x):
    if x.max() == x.min():
        return np.zeros_like(x, dtype=float)
    return (x - x.min()) / (x.max() - x.min())

t_norm = normalizar(tiempos)
c_norm = normalizar(costos)
r_norm = normalizar(riesgos)

# 4) Utilidad (más alto = mejor)
utilidades = {}
nombres = list(rutas.keys())

for i, nombre in enumerate(nombres):
    penalizacion = (w_tiempo * t_norm[i]) + (w_costo * c_norm[i]) + (w_riesgo * r_norm[i])
    utilidad = 1 - penalizacion
    utilidades[nombre] = utilidad

# 5) Elegimos la mejor
mejor_ruta = max(utilidades, key=utilidades.get)

print("Utilidades:")
for k, v in utilidades.items():
    print(f"  Ruta {k}: utilidad={v:.3f} | datos={rutas[k]}")

print("\nMejor decisión según utilidad:", mejor_ruta)
```

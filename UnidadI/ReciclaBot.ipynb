Se observan **fotos de vertederos locales** y se plantea el reto:

## 📦 Librerías utilizadas
📌 Usamos `random` para simular un entorno impredecible, como ocurre en la realidad.


```python
import random
```

## 👁️ Sensores: Percepción del entorno

### Sensor de cámara (visión artificial simulada)
🔎 **¿Qué representa?**
- Una cámara industrial
- Un sistema de visión artificial
- Un modelo de reconocimiento de objetos


```python
def sensor_camara():
    # Simulamos que la cámara detecta un material al azar
    materiales = ["botella_pet", "lata_aluminio", "carton_pizza", "manzana"]
    return random.choice(materiales)
```

### Sensor de peso (báscula)
⚖️ **Idea clave**:
> Una IA real **combina múltiples sensores** para tomar mejores decisiones.


```python
def sensor_peso(material):
    # Simulamos el peso en gramos
    pesos = {
        "botella_pet": 20, 
        "lata_aluminio": 15, 
        "carton_pizza": 100, 
        "manzana": 150
    }
    return pesos.get(material, 0)
```

## 🤖 El Agente Inteligente (Cerebro)
El agente sigue el ciclo clásico de la IA:

1. **Percibir**
2. **Decidir**
3. **Actuar**

📌 **Tipo de agente:**
> Agente **reactivo simple** (no tiene memoria ni aprendizaje)


```python
def agente_reciclador():
    print("--- INICIANDO CICLO DE RECICLAJE ---")

    # 1. PERCEPCIÓN (SENSING)
    objeto_visto = sensor_camara()
    peso_detectado = sensor_peso(objeto_visto)
    print(f"👁️ Detectado: {objeto_visto} ({peso_detectado}g)")

    # 2. DECISIÓN (THINKING)
    accion = ""

    # Lógica Reactiva: Condición → Acción
    if objeto_visto == "botella_pet" or objeto_visto == "lata_aluminio":
        accion = "MOVER_A_CONTENEDOR_RECICLABLE"
    elif objeto_visto == "manzana":
        accion = "MOVER_A_COMPOSTA"
    else:
        accion = "MOVER_A_BASURA_GENERAL"

    # 3. ACTUACIÓN (ACTING)
    print(f"🤖 Acción ejecutada: {accion}\n")
```

## 🔁 Ejecución del sistema en tiempo real
Simulamos una banda transportadora con varios objetos:

🧠 **Interpretación**:
- Cada ciclo representa un objeto en la banda
- El agente decide de forma autónoma


```python
# Ejecución del ciclo 5 veces
for _ in range(5):
    agente_reciclador()
```

## ❓ Pregunta de reflexión
> **¿Qué pasa si la botella de PET tiene líquido dentro y pesa más?**

🔴 **Problema**:
- El agente clasifica solo por tipo
- Ignora el contexto del peso
- Puede cometer errores


## 🧩 Conclusión

✔️ Este agente:
- Es autónomo
- Percibe el entorno
- Toma decisiones

❌ Pero:
- No recuerda experiencias pasadas
- No aprende
- No se adapta

📌 **Conclusión clave**:
> *Los agentes reactivos simples son limitados. Para mejorar, necesitamos memoria, estado interno o modelos de aprendizaje.*
